## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Replacing NaN values with "Not Rated" in the age_certification column
title_df['age_certification'] = title_df['age_certification'].fillna('Not Rated')

# Filling null values with mean values in imdb_score column
title_df['imdb_score'].fillna(title_df['imdb_score'].mean(), inplace=True)

# Dropping duplicate values for both title_df and credit_df
title_df.drop_duplicates(inplace=True)
credit_df.drop_duplicates(inplace=True)

# Merging title_df dataset and credit_df dataset using 'id' as primary key
merged_df = title_df.merge(credit_df, on='id', how='left')

# Creating a column for decade
title_df['decade'] = (title_df['release_year'] // 10) * 10

# Categorize runtime into short, medium, long
merged_df['runtime_category'] = ''
for i in range(len(merged_df)):
    if merged_df.loc[i, 'runtime'] < 30:
        merged_df.loc[i, 'runtime_category'] = 'Short'
    elif merged_df.loc[i, 'runtime'] <= 90:
        merged_df.loc[i, 'runtime_category'] = 'Medium'
    else:
        merged_df.loc[i, 'runtime_category'] = 'Long'

# Add column is_movie (1 if movie, 0 if show)
merged_df['is_movie'] = 0
for i in range(len(merged_df)):
    if merged_df.loc[i, 'type'] == 'MOVIE':
        merged_df.loc[i, 'is_movie'] = 1
    else:
        merged_df.loc[i, 'is_movie'] = 0
```

```python
for i in range(len(merged_df)):
    score = merged_df.loc[i, 'imdb_score']

    if pd.isna(score):
        merged_df.loc[i, 'rating_category'] = 'not rated'

    elif score < 5:
        merged_df.loc[i, 'rating_category'] = 'poor'

    elif score < 7:
        merged_df.loc[i, 'rating_category'] = 'average'

    elif score < 8.5:
        merged_df.loc[i, 'rating_category'] = 'good'

    else:
        merged_df.loc[i, 'rating_category'] = 'excellent'


```

### What all manipulations have you done and insights you found?

I cleaned the data by handling missing values and duplicates, correcting data types, and performing filtering, grouping, and visualization. I found insights about popular genres, content types, ratings, countries, and yearly content trends.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=title_df, y='imdb_score')
plt.title('box plot of imdb_score\n', color = 'brown')
plt.show()
```

2. There are **outliers on both ends** — a cluster of titles scoring **above 9** (highly rated gems) and a cluster scoring **below 3** (poorly rated content).

```python
#Blox plot for runtime
plt.figure(figsize=(8,6))
sns.boxplot(data=title_df, x='runtime')
plt.title('box plot of runtime\n', color = 'brown')
plt.show()
```

INSIGHTS
